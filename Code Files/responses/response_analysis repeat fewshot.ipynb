```python
import pandas as pd
api = pd.read_csv("../responses/responses_raw.csv")
api.head()
```

```python
man = pd.read_csv("../responses/responses_manual.csv")
man.head()
```

```python
few_shot = pd.read_csv("../responses/few_shot_only_responses_raw.csv")
few_shot.head()
```

```python
pd.set_option('display.max_colwidth', 100)
comb_df = pd.concat([api,man,few_shot],ignore_index=True)
comb_df.head()
```

```python
comb_df['query_mode'].value_counts()
```

```python
comb_df.shape
```

```python
comb_df['model'].value_counts()
```

```python
#Creating a subset of only the columns needed
sub = (
    comb_df[['sentence','query_mode','correct','context_type']]
    .copy()
    .assign(
        prompt_type=comb_df['prompt_id'].str.split('_').str[0],
        shot_type=comb_df['prompt_id'].str.split('_').str[1],
        model=comb_df['model'].str.split('-').str[0]
    )
)
new_order = ['sentence','model','query_mode','prompt_type',
             'shot_type','context_type','correct']
sub = sub[new_order]
sub.head()
```

```python
sub.to_csv("../responses/mixed-effects.csv",index = False)
```
