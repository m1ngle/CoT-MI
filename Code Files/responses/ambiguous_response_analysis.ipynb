```python
#Function to run McNemar Tests
from statsmodels.stats.contingency_tables import mcnemar

def run_mcnemar(df, colA, colB):

    tmp = df[[colA, colB]].dropna().copy()

    table = pd.crosstab(tmp[colA], tmp[colB])
    table = table.reindex(index=[False, True], columns=[False, True], fill_value=0)

    b = table.loc[True, False]
    c = table.loc[False, True]
    n = b + c
#adding different tests per number discordant pairs
    if n <= 25:
        result = mcnemar(table, exact=True)
        test_type = "exact"
    else:
        result = mcnemar(table, exact=False, correction=True)
        test_type = "asymptotic"

    return table, result.pvalue, n, test_type
```

```python
# Comparison conditions
comparison_specs = {
    "CoT vs Direct": [
        ("api_cot_zero", "api_direct_zero"),
        ("api_cot_few", "api_direct_few"),
        ("manual_cot_zero", "manual_direct_zero"),
        ("manual_cot_few", "manual_direct_few"),
    ],
    "Few vs Zero": [
        ("api_cot_few", "api_cot_zero"),
        ("api_direct_few", "api_direct_zero"),
        ("manual_cot_few", "manual_cot_zero"),
        ("manual_direct_few", "manual_direct_zero"),
    ],
    "Manual vs API": [
        ("manual_cot_zero", "api_cot_zero"),
        ("manual_cot_few", "api_cot_few"),
        ("manual_direct_zero", "api_direct_zero"),
        ("manual_direct_few", "api_direct_few"),
    ],
    "CoT Few vs CoT Zero": [
        ("api_cot_few", "api_cot_zero"),
        ("manual_cot_few", "manual_cot_zero"),
    ],
    "Direct Few vs Direct Zero": [
        ("api_direct_few", "api_direct_zero"),
        ("manual_direct_few", "manual_direct_zero"),
    ],
}

results = []
```

```python
print(sub_wide_df['manual_cot_zero'].value_counts(), sub_wide_df['manual_direct_zero'].value_counts())
```

```python
for model_name in sub_wide_df["model"].dropna().unique():
    df_model = sub_wide_df[sub_wide_df["model"] == model_name].copy()

    for comparison_name, pairs in comparison_specs.items():
        for colA, colB in pairs:
            table, pvalue, n, test_type = run_mcnemar(df_model, colA, colB)

            # Safely extract 2x2 counts even if some combinations are absent
            a = table.loc[True, True] if True in table.index and True in table.columns else 0
            b = table.loc[True, False] if True in table.index and False in table.columns else 0
            c = table.loc[False, True] if False in table.index and True in table.columns else 0
            d = table.loc[False, False] if False in table.index and False in table.columns else 0

            results.append({
                "model": model_name,
                "comparison": comparison_name,
                "condition_A": colA,
                "condition_B": colB,
                "both_correct": a,
                "A_correct_B_wrong": b,
                "A_wrong_B_correct": c,
                "both_wrong": d,
                "p_value": pvalue
            })

results_df = pd.DataFrame(results)
results_df.head()
```

```python
#Changing McNemar to check for robustness
from statsmodels.stats.contingency_tables import mcnemar

def run_mcnemar(df, colA, colB):

    tmp = df[[colA, colB]].dropna().copy()

    table = pd.crosstab(tmp[colA], tmp[colB])
    table = table.reindex(index=[False, True], columns=[False, True], fill_value=0)

    b = table.loc[True, False]
    c = table.loc[False, True]
    n = b + c
#adding different tests per number discordant pairs
    if n <= 25:
        result = mcnemar(table, exact=False, correction=True)
        test_type = "exact"
    else:
        result = mcnemar(table, exact=False, correction=True)
        test_type = "asymptotic"

    return table, result.pvalue, n, test_type
```

```python
model_wide = pd.pivot_table(
    sub_dedup,
    values='correct',
    index=['sentence',  'shot','query_mode','prompt_style'],
    columns= 'model',
    aggfunc='first'
).reset_index()
```

```python
from statsmodels.stats.contingency_tables import cochrans_q, mcnemar


def run_model_comparison(df, *model_cols):
    """
    Run Cochran's Q test followed by pairwise McNemar tests.
    
    Parameters:
    df: DataFrame with model performance columns
    *model_cols: column names for each model (e.g., 'gpt', 'deepseek', 'claude')
    
    Returns:
    overall: DataFrame with Cochran's Q results
    pairwise: DataFrame with pairwise McNemar comparisons
    """
    # Keep only complete cases
    tmp = df[list(model_cols)].dropna().copy()
    
    # Overall Cochran's Q test
    q_result = cochrans_q(tmp)
    
    # Calculate accuracies
    accuracies = tmp.mean()
    
    # Overall result
    overall = pd.DataFrame([{
        "test": "Cochran's Q",
        "statistic": q_result.statistic,
        "p_value": q_result.pvalue,
        "df": len(model_cols) - 1,
        "n_observations": len(tmp),
        **{f"{col}_accuracy": accuracies[col] for col in model_cols}
    }])
    
    # Pairwise McNemar tests (post-hoc)
    pairwise_results = []
    n_comparisons = len(list(model_cols)) * (len(list(model_cols)) - 1) // 2
    bonferroni_alpha = 0.05 / n_comparisons
    
    for i, colA in enumerate(model_cols):
        for colB in model_cols[i+1:]:
            table = pd.crosstab(tmp[colA], tmp[colB])
            table = table.reindex(index=[False, True], columns=[False, True], fill_value=0)
            
            b = table.loc[True, False]
            c = table.loc[False, True]
            n = b + c
            
            if n <= 25:
                mcnemar_result = mcnemar(table, exact=True)
                test_type = "exact"
            else:
                mcnemar_result = mcnemar(table, exact=False, correction=True)
                test_type = "asymptotic"
            
            pairwise_results.append({
                "comparison": f"{colA} vs {colB}",
                "both_correct": table.loc[True, True],
                f"{colA}_correct_{colB}_wrong": table.loc[True, False],
                f"{colA}_wrong_{colB}_correct": table.loc[False, True],
                "both_wrong": table.loc[False, False],
                "discordant_pairs": n,
                "test_type": test_type,
                "p_value": mcnemar_result.pvalue,
                "bonferroni_alpha": bonferroni_alpha,
                "significant": mcnemar_result.pvalue < bonferroni_alpha
            })
    
    pairwise_df = pd.DataFrame(pairwise_results)
    
    return overall, pairwise_df

# Usage:
overall_result, pairwise_results = run_model_comparison(model_wide, 'gpt', 'deepseek', 'claude')

print("="*60)
print("OVERALL TEST: Cochran's Q")
print("="*60)
print(overall_result)

print("\n" + "="*60)
print("POST-HOC PAIRWISE COMPARISONS (with Bonferroni correction)")
print("="*60)
print(pairwise_results)

# Interpretation
if overall_result['p_value'].values[0] < 0.05:
    print("\n✓ Cochran's Q significant: Models differ in performance")
    print(f"  Bonferroni-corrected alpha: {pairwise_results['bonferroni_alpha'].values[0]:.4f}")
    sig_pairs = pairwise_results[pairwise_results['significant']]
    if len(sig_pairs) > 0:
        print(f"\n  Significant pairwise differences:")
        for _, row in sig_pairs.iterrows():
            print(f"    - {row['comparison']}: p = {row['p_value']:.4f}")
else:
    print("\n✗ Cochran's Q not significant: No evidence of model differences")
```
