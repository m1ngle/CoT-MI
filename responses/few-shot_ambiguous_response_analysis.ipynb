```python
from pathlib import Path
import pandas as pd

# Get the notebook's directory
notebook_dir = Path.cwd()
file_path = notebook_dir / "ambiguous_responses_raw.csv"
```

```python
# A smaller subsection of 156 responses were generated for manual and API calls (total 312).
# These sentences were more ambiguous and designed to suggest more than one
# The responses below are API calls
aapi = pd.read_csv(file_path)
aapi.head()
```

```python
# 7 responses in API call were cut due to length constraint of 300 tokens. 
# These responses will be dropped, with tokens increased to 500 to gather
# full responses.
aapi_drop = aapi.dropna()
aapi_drop.isna().sum()
```

```python
laapi = pd.read_csv("long_ambiguous_responses_raw.csv")
laapi.head()
```

```python
# 156 manual responses from
aman = pd.read_excel("manual_ambiguous_responses_raw.xlsm")
aman.head()
```

```python
few_am = pd.read_csv("ambiguous_few_shot_only_responses_raw.csv")
few_am.head()
```

```python
few_am.isna().sum()
```

```python
# One response in API call was cut due to length constraint of 300 tokens. 
# This responses will be dropped, with tokens increased to 500 to gather
# full response.
few_am_drop = few_am.dropna()
few_am_drop.isna().sum()
```

```python
few_am_drop.shape
```

```python
rerun_few=pd.read_csv("rerun_ambiguous_few_shot_only_responses_raw.csv")
rerun_few
```

```python
pd.set_option('display.max_colwidth', 100)
aapi_man = pd.concat([aapi_drop,aman,laapi,few_am_drop,rerun_few],ignore_index=True)
aapi_man.head()
```

```python
aapi_man['query_mode'].value_counts()
```

```python
aapi_man.shape
```

```python
aapi_man['model'].value_counts()
```

```python
aapi_man['correct'].value_counts()
```

```python
#Creating a subset of only the columns needed
sub = (
    aapi_man[['sentence','query_mode','correct','context_type']]
    .copy()
    .assign(
        prompt_type=aapi_man['prompt_id'].str.split('_').str[0],
        shot_type=aapi_man['prompt_id'].str.split('_').str[1],
        model=aapi_man['model'].str.split('-').str[0]
    )
)
new_order = ['sentence','model','query_mode','prompt_type',
             'shot_type','context_type','correct']
sub = sub[new_order]
sub.head()
```

```python
sub.to_csv("../responses/ambiguous_mixed-effects.csv",index = False)
```
